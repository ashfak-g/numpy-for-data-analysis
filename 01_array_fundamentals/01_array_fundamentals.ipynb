**Data Type & Memory Optimization**

```python
# Convert the dataset numbers to float32 to save memory space
data_float32 = data.astype(np.float32)
```

```python
# Print the original data type of our main dataset
print(data.dtype)
# Print the data type of our newly converted dataset
print(data_float32.dtype)
```

```python
# Print the memory size in bytes of the original dataset
print(data.nbytes)
# Print the memory size in bytes of the converted dataset
print(data_float32.nbytes)
```

```python
# Get only the last column of data and convert it to tiny integers
quality = data[:, -1].astype(np.int8)
# Print the data type of this quality column
print(quality.dtype)
# Print the tiny memory size used by the quality column
print(quality.nbytes)
```

**Python List vs NumPy Array**

```python
import numpy as np
import sys

# Make a list with numbers
my_list = [10, 20, 30, 40]

# Make an array with the same numbers
my_array = np.array([10, 20, 30, 40])

# 1. Show the data types
print(type(my_list))
print(type(my_array))

# 2. Show math differences
print(my_list * 2)
print(my_array * 2)

# 3. Show memory use
list_mem = sys.getsizeof(my_list) + sum(sys.getsizeof(i) for i in my_list)
array_mem = my_array.nbytes
print(list_mem)
print(array_mem)

# 4. Show same type rule
mixed_list = [10, "hello", 3.14]
print(mixed_list)

mixed_array = np.array([10, "hello", 3.14])
print(mixed_array)
```
